```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.read_csv("/content/amazon_sales_data 2025.csv")
```

```python
az=pd.read_csv("/content/amazon_sales_data 2025.csv")
```

```python
az.info()
```

```python
az.head(50)
```

```python
az.isna().sum()
```

```python
az.duplicated().sum()
```

```python
az.groupby("Category")['Total Sales'].sum()
```

```python
sns.barplot(x="Category",y="Total Sales",data=az,hue='Category')
plt.title("Total Sales by Category")
plt.grid('axis=y')
plt.xticks(rotation=45)
plt.show()
```

```python
az.groupby("Product")['Total Sales'].mean()
```

```python
sns.violinplot(x="Product",y="Total Sales",data=az,hue='Product')
plt.title("Total Sales by Product")
plt.grid('axis=y')
plt.xticks(rotation=45)
plt.show()
```

```python
az.groupby("Product")['Status'].value_counts()
```

```python
sns.countplot(x="Product", hue="Status", data=az)
plt.title("Status by Product")
plt.grid(axis='y')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
sns.barplot(x="Category", y="Total Sales", data=az, hue='Category')
plt.title("Total Sales by Category")
plt.grid(axis='y')
plt.xticks(rotation=45)

plt.subplot(1, 3, 2)
sns.violinplot(x="Product", y="Total Sales", data=az, hue='Product')
plt.title("Total Sales by Product")
plt.grid(axis='y')
plt.xticks(rotation=45)


plt.subplot(1, 3, 3)
sns.countplot(x="Product", hue="Status", data=az)
plt.title("Status by Product")
plt.grid(axis='y')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
